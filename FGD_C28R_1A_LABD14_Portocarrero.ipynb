### Los roles del gobierno: DAMA-DMBOK

Ojo con esto, porque es la confusión más común del curso. **Hay dos tipos de rol y no son lo mismo:**

| | **Roles operativos** (los tuyos, desde la S11) | **Roles de gobierno** (DAMA) |
|---|---|---|
| Quiénes | Vendedor, Cajero, Supervisor, Auditor | Data Owner, Data Steward, Data Engineer, Data Analyst |
| Qué definen | Qué **puede hacer** en el sistema | Quién **responde** por el dato |
| Ejemplo | El vendedor puede insertar | El Data Owner **decide** que el vendedor pueda insertar |

Los roles operativos son **permisos**. Los de gobierno son **responsabilidades**. Un permiso se revoca con una línea de código; una responsabilidad, no.

| Rol DAMA | Qué hace | En MediSur |
|---|---|---|
| **Data Owner** | **Decide.** Es el dueño del dato y responde por él ante la ley | El gerente: decide quién ve `condicion_cronica` |
| **Data Steward** | **Cuida.** Vigila la calidad y define las reglas del día a día | El jefe de sede: revisa que los DNI sean válidos |
| **Data Engineer** | **Construye.** Arma los pipelines y las bases | Tú, cuando hiciste el ETL de la S10 |
| **Data Analyst** | **Interpreta.** Convierte el dato en decisiones | Quien arma el reporte del gerente |

La pregunta que revela quién es el Data Owner es siempre la misma: **si estos datos se filtran mañana, ¿a quién despiden?** Esa persona es el Owner. No es un cargo técnico: es quien firma.

### Privacidad: la ley no es opcional

- **Ley 29733 (Perú)**: datos sensibles son salud, **ingresos económicos**, origen racial y étnico, convicciones políticas o religiosas, afiliación sindical y biométricos.
- **GDPR (Europa)**: derecho al olvido, consentimiento explícito, multas de hasta el 4% de la facturación global.

Las dos exigen lo mismo que vas a construir hoy: **saber quién accedió a qué, y poder demostrarlo**. Sin bitácora, no hay cumplimiento. No importa lo bien intencionado que seas: si no puedes demostrarlo, no lo hiciste.

```python
# ============================================================
# Paso 1: recuperar el proyecto (semanas 10 a 13)
# ============================================================
import sqlite3, os, re, datetime
import pandas as pd
import numpy as np

caso = "03_farmacia_medisur"          # <-- cambia por la carpeta de TU caso asignado
DB_ORIGEN  = "farmacia_legacy.db"
DB_DESTINO = "farmacia_migrada.db"

REPO = "Rociosayan/-PMD2_FGD_Bases_Oficiales_FDG"
url  = f"https://raw.githubusercontent.com/{REPO}/main/casos/{caso}/{caso}.db"
TABLAS_DEL_SISTEMA = {"catalogo_metadatos", "ficha_dublin_core", "resumen_etl",
                      "reglas_calidad", "ventas_para_revisar", "auditoria"}


def descargar():
    if os.path.exists(DB_ORIGEN):
        return
    import requests
    r = requests.get(url)
    if r.status_code == 200 and r.content[:16] == b"SQLite format 3\x00":
        open(DB_ORIGEN, "wb").write(r.content); print("Base oficial descargada.")
    else:
        from google.colab import files
        print("Sube manualmente el .db de tu caso:")
        s = files.upload(); open(DB_ORIGEN, "wb").write(list(s.values())[0])


def migracion_semana10():
    o = sqlite3.connect(DB_ORIGEN); d = sqlite3.connect(DB_DESTINO)
    cli = pd.read_sql_query("SELECT * FROM clientes", o).drop_duplicates(
        subset=["num_documento"], keep="first")
    for col in ["nombres", "apellidos", "distrito", "segmento", "tipo_documento"]:
        cli[col] = cli[col].astype("string").str.strip().str.title()
    ok = cli["correo"].str.match(r"^[^@\s]+@[^@\s]+\.[^@\s]+$", na=False)
    cli["correo_fue_corregido"] = cli["correo"].notna() & ~ok
    cli.loc[~ok, "correo"] = pd.NA
    cli["correo"]   = cli["correo"].fillna("SIN CORREO REGISTRADO")
    cli["telefono"] = cli["telefono"].fillna("000000000")
    cli.to_sql("clientes_limpio", d, if_exists="replace", index=False)
    op = pd.read_sql_query("SELECT * FROM operaciones", o)
    op["monto_ajustado"]     = np.where(op["monto_total"] < 0, 0, op["monto_total"])
    op["monto_fue_ajustado"] = op["monto_total"] < 0
    op.to_sql("ventas_limpia", d, if_exists="replace", index=False)
    ya = set(pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table'", d)["name"])
    for t in ["sedes", "empleados", "productos_servicios", "detalle_operacion",
              "pagos", "incidencias"]:
        if t not in ya:
            pd.read_sql_query(f"SELECT * FROM {t}", o).to_sql(t, d, if_exists="replace", index=False)
    d.commit(); o.close(); d.close()


def catalogo_semana11():
    c = sqlite3.connect(DB_DESTINO)
    tablas = [t for t in pd.read_sql_query(
        "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name", c)["name"]
        if t not in TABLAS_DEL_SISTEMA]
    filas = []
    for tabla in tablas:
        for _, col in pd.read_sql_query(f"PRAGMA table_info({tabla})", c).iterrows():
            filas.append({"tabla": tabla, "columna": col["name"],
                          "tipo_dato": col["type"] or "SIN TIPO"})
    cat = pd.DataFrame(filas)
    DELICADO = {"01_clinica_vitalsalud":("diagnostico","Sensible"),
        "02_banco_crediandes":("saldo_cuenta","Sensible"),
        "03_farmacia_medisur":("condicion_cronica","Sensible"),
        "04_vet_cuidapatas":("motivo_atencion_mascota","Interno"),
        "05_hotel_andes_stay":("nacionalidad","Confidencial"),
        "06_logiexpress_urbano":("direccion_entrega","Confidencial"),
        "07_universidad_aula360":("promedio_notas","Confidencial"),
        "08_muni_data":("tipo_tramite","Interno"),
        "09_retail_online_market":("tarjeta_terminacion","Confidencial"),
        "10_agroexport_sol":("pais_destino","Interno"),
        "11_energyhome_servicios":("consumo_kwh","Confidencial"),
        "12_seguros_protege":("monto_asegurado","Sensible"),
        "13_fastfood_norte":("direccion_delivery","Confidencial"),
        "14_manufactura_metaltec":("linea_credito","Sensible"),
        "15_telconorte_movil":("numero_linea","Confidencial")}
    SENS = {"num_documento":"Confidencial","correo":"Confidencial","telefono":"Confidencial",
            "nombres":"Interno","apellidos":"Interno","nombre_producto":"Publico",
            "categoria":"Publico","precio_base":"Publico","activo":"Publico"}
    col_caso, niv = DELICADO[caso]; SENS[col_caso] = niv
    cat["sensibilidad"] = cat["columna"].map(lambda x: SENS.get(x, "Interno"))
    cat.to_sql("catalogo_metadatos", c, if_exists="replace", index=False)
    c.commit(); c.close()
    return col_caso


descargar()
if not os.path.exists(DB_DESTINO):
    migracion_semana10(); print("Semana 10 reconstruida.")
conn = sqlite3.connect(DB_DESTINO)
hay = lambda t: len(pd.read_sql_query(
    f"SELECT name FROM sqlite_master WHERE type='table' AND name='{t}'", conn)) > 0
if not hay("catalogo_metadatos"):
    conn.close(); catalogo_semana11(); conn = sqlite3.connect(DB_DESTINO)
    print("Semana 11 reconstruida.")

# Las semanas 11 y 12: acceso de LECTURA gobernado por el catalogo
NIVEL_ACCESO = {"Vendedor":{"Publico","Interno"}, "Cajero":{"Publico","Interno"},
                "Supervisor":{"Publico","Interno","Confidencial"},
                "Auditor":{"Publico","Interno","Confidencial","Sensible"},
                "DataOwner":{"Publico","Interno","Confidencial","Sensible"}}

def columnas_visibles(tabla, rol):
    permitidos = NIVEL_ACCESO.get(rol, set())
    if not permitidos: return []
    cat = pd.read_sql_query(
        "SELECT columna, sensibilidad FROM catalogo_metadatos WHERE tabla = ?",
        conn, params=(tabla,))
    return [c for c, s in zip(cat["columna"], cat["sensibilidad"]) if s in permitidos]

def ver_clientes(rol, limite=8):
    cols = columnas_visibles("clientes_limpio", rol)
    if not cols: return f"ACCESO DENEGADO: el rol {rol} no puede consultar."
    return pd.read_sql_query(f"SELECT {', '.join(cols)} FROM clientes_limpio LIMIT {limite}", conn)

print("Proyecto recuperado. Semanas 10 a 13 disponibles.")
print("Hasta hoy tu sistema solo sabe LEER. El Vendedor ve",
      len(columnas_visibles("clientes_limpio", "Vendedor")), "columnas y no puede tocar nada.")
```

### Paso 2: la matriz de gobierno — quién responde por cada dato

Antes de dar permisos, hay que decidir **quién responde**. Esta es la tabla que el gerente firma.

Fíjate en la columna `data_owner`: no dice "el sistema" ni "TI". Dice **una persona**. Un dato sin dueño es un dato del que nadie responde, y eso es exactamente lo que la Ley 29733 no perdona.

```python
# ============================================================
# Paso 2: matriz de gobierno DAMA sobre los campos criticos
# ============================================================
# Tomamos las columnas criticas que el catalogo de la S11 ya marco.
criticas = pd.read_sql_query("""
    SELECT tabla, columna, sensibilidad FROM catalogo_metadatos
    WHERE sensibilidad IN ('Sensible', 'Confidencial')
    ORDER BY CASE sensibilidad WHEN 'Sensible' THEN 1 ELSE 2 END, tabla, columna""", conn)

# Quien RESPONDE por cada tipo de dato. Esto no lo decide el programador: lo firma el gerente.
GOBIERNO = {
    "Sensible":     {"data_owner": "Gerente General",
                     "data_steward": "Jefe de Cumplimiento",
                     "base_legal": "Ley 29733 - dato sensible: proteccion reforzada"},
    "Confidencial": {"data_owner": "Jefe de Sede",
                     "data_steward": "Supervisor de Atencion",
                     "base_legal": "Ley 29733 - dato personal: consentimiento del titular"},
}

matriz = criticas.copy()
matriz["data_owner"]   = matriz["sensibilidad"].map(lambda s: GOBIERNO[s]["data_owner"])
matriz["data_steward"] = matriz["sensibilidad"].map(lambda s: GOBIERNO[s]["data_steward"])
matriz["base_legal"]   = matriz["sensibilidad"].map(lambda s: GOBIERNO[s]["base_legal"])
matriz["data_engineer"] = "Equipo de datos (tu)"

matriz.to_sql("matriz_gobierno", conn, if_exists="replace", index=False)
conn.commit()

print("MATRIZ DE GOBIERNO DE DATOS")
print("=" * 60)
print(f"Campos criticos bajo gobierno: {len(matriz)}")
display(matriz[["tabla", "columna", "sensibilidad", "data_owner", "data_steward"]])
print()
print("La pregunta que revela al Data Owner:")
print("   'Si estos datos se filtran manana, a quien despiden?'")
print("   Esa persona es el Owner. No es un cargo tecnico: es quien firma.")
```

**Pregunta 1:** Mira la matriz. ¿Por qué el `data_owner` de un dato **Sensible** es el Gerente General y no el jefe de sede? Y una pregunta más difícil: **¿por qué el Data Engineer (tú) no es dueño de ningún dato**, si eres quien construyó toda la base?

Respuesta: Porque el tratamiento de datos sensibles, es un proceso que atañe a toda una organización y no sólo a una sede. Los datos sensibles, al tener una protección reforzada, necesitan de la máxima autoridad en un titular de banco de datos personales para determinar su contenido, tratamiento y seguridad, de forma apropiada y acorde con la categoría de datos que se trate (Ley Nº 29733, Título Preliminar, art. 17 y Título I, art. 9).

Un Data Engineer no puede ser dueño de ningún dato (más allá de la titularidad sobre sus propios datos personales), porque su responsabilidad es de naturaleza operativa y no legal. Un Data Engineer realiza sus labores de contrucción y mantenimiento de infraestructura según las políticas dadas por el Data Owner, pero no tiene poder de decisión sobre los datos. Se trata de un encargado de tratamiento de los datos, de acuerdo a la Ley 29733 y un Data Custodian en el marco de DAMA International.

### Paso 3: permisos de escritura — el sistema empieza a poder hacer daño

Hasta ahora un rol solo definía **qué columnas ves**. Desde hoy define **qué operaciones puedes hacer**: `SELECT`, `INSERT`, `UPDATE`, `DELETE`. Son las cuatro operaciones del **DML** que viste en la Semana 9.

Mira bien el reparto, porque cada línea es una decisión de negocio, no técnica:

| Rol | Puede | Por qué |
|---|---|---|
| **Vendedor** | `INSERT` | Da de alta clientes y registra ventas. Si se equivoca, se corrige |
| **Cajero** | `UPDATE` | Corrige datos de contacto. No crea ni destruye |
| **Supervisor** | `DELETE` | Es el **único** que borra. Borrar es irreversible: se concentra en una persona |
| **Auditor** | `SELECT` | **Solo mira.** Quien fiscaliza no puede modificar lo que fiscaliza |
| **Data Owner** | todo | Responde por el dato, así que puede todo |

Fíjate en el Auditor: es el rol con **más acceso de lectura** (ve hasta lo Sensible) y **cero** de escritura. No es un descuido: es el principio de que **nadie audita lo que puede modificar**.

> **Nota sobre SQLite:** PostgreSQL o SQL Server harían esto con `GRANT` y `REVOKE` —el lenguaje **DCL**—. SQLite no tiene permisos reales, así que los simulamos en Python. El concepto es idéntico; cambia la herramienta.

```python
# ============================================================
# Paso 3: permisos de ESCRITURA (DCL simulado)
# ============================================================
PERMISOS = {
    "Vendedor":   {"SELECT", "INSERT"},           # da de alta clientes y registra ventas
    "Cajero":     {"SELECT", "UPDATE"},           # corrige datos de contacto
    "Supervisor": {"SELECT", "DELETE"},           # el UNICO que puede borrar
    "Auditor":    {"SELECT"},                     # solo mira: nadie audita lo que puede modificar
    "DataOwner":  {"SELECT", "INSERT", "UPDATE", "DELETE"},
}


def puede(rol, operacion):
    """El permiso de ESCRITURA. Complementa a columnas_visibles(), que es de lectura."""
    return operacion in PERMISOS.get(rol, set())


# La matriz que el gerente firma
filas = []
for rol, ops in PERMISOS.items():
    filas.append({"rol": rol,
                  "SELECT": "Si" if "SELECT" in ops else "-",
                  "INSERT": "Si" if "INSERT" in ops else "-",
                  "UPDATE": "Si" if "UPDATE" in ops else "-",
                  "DELETE": "Si" if "DELETE" in ops else "-",
                  "ve_datos_sensibles": "Si" if "Sensible" in NIVEL_ACCESO[rol] else "-"})
matriz_permisos = pd.DataFrame(filas)
matriz_permisos.to_sql("matriz_permisos", conn, if_exists="replace", index=False)
conn.commit()

print("MATRIZ DE PERMISOS POR ROL")
print("=" * 60)
display(matriz_permisos)
print("Vendedor puede INSERT?  ", puede("Vendedor", "INSERT"))
print("Vendedor puede DELETE?  ", puede("Vendedor", "DELETE"))
print("Auditor  puede UPDATE?  ", puede("Auditor",  "UPDATE"))
print()
print("Fijate en el AUDITOR: ve TODO (hasta lo Sensible) y no puede escribir NADA.")
print("Nadie audita lo que puede modificar.")
```

**Pregunta 2:** El Auditor es el rol que **más datos ve** y el que **menos puede tocar**. Explica por qué eso no es una contradicción sino el punto. Luego: **¿por qué solo el Supervisor puede `DELETE`?** ¿Qué pasaría si el vendedor también pudiera?

Respuesta: Puesto que el propósito del trabajo de un auditor es revisar el estado actual de una organización para comprobar si cumple las leyes, tiene todo el sentido que no pueda ni deba modificar nada pero sí ver todo. Además se debe de tener en cuenta que muchas veces las auditorías son externas.

Un supervisor puede leer y borrar registros porque es parte de su labor de mantener la calidad en el marco de la gobernanza de datos, detectando y corrigiendo errores que dañen los tableros de análisis o ya no aporten información útil, lo cual se apega al principio de mínimo privilegio. Si un vendedor pudiera borrar información, se rompería el principio mencionado anteriormente, y expondría a la empresa a pérdida de información relevante por falta de competencias o por intencionalidad.

### Paso 4: validar antes de escribir — el portero de la base

Aquí está el momento más importante del laboratorio, y casi nadie lo ve venir.

Llevas cuatro semanas limpiando esta base. La Semana 13 la dejó en 92 sobre 100. Y hoy vas a abrir la puerta para que **la gente escriba directamente en ella**.

Si el vendedor teclea `"Juan123"` como nombre, o `"correo.com"` sin arroba, o un DNI de 5 dígitos, **acabas de meter a mano la misma suciedad que llevas un mes sacando**. Tu score se desploma y esta vez la culpa es tuya, porque tenías la oportunidad de impedirlo.

Las validaciones son ese portero. Y fíjate en la conexión: **cada validación de abajo es una regla de calidad de la Semana 13, pero aplicada antes de grabar en vez de después**.

> Limpiar es curar. Validar es **prevenir**. La calidad no se arregla al final del pipeline: se defiende en la puerta de entrada.

```python
# ============================================================
# Paso 4: validaciones -- las reglas de la S13, aplicadas ANTES de grabar
# ============================================================
def validar_nombre(nombre):
    """Un nombre no lleva numeros. Regla de EXACTITUD (S13)."""
    if not nombre or not str(nombre).strip():
        return "No puede estar vacio."
    if not str(nombre).replace(" ", "").isalpha():
        return "Solo debe contener letras."
    return None


def validar_dni(dni):
    """Un DNI peruano tiene EXACTAMENTE 8 digitos. Es la regla que la S13 detecto rota."""
    dni = str(dni).strip()
    if not dni:            return "El DNI no puede estar vacio."
    if not dni.isdigit():  return "El DNI solo debe contener numeros."
    if len(dni) != 8:      return f"El DNI debe tener 8 digitos (tiene {len(dni)})."
    return None


def validar_correo(correo):
    """Regla de EXACTITUD: el correo debe tener formato valido."""
    if not correo or not str(correo).strip():
        return "El correo no puede estar vacio."
    if not re.match(r"^[^@\s]+@[^@\s]+\.[^@\s]+$", str(correo)):
        return "El correo no es valido (ej: nombre@dominio.com)."
    return None


def validar_telefono(tel):
    tel = str(tel).strip()
    if not tel:            return "El telefono no puede estar vacio."
    if not tel.isdigit():  return "El telefono solo debe contener numeros."
    if len(tel) != 9:      return f"El telefono debe tener 9 digitos (tiene {len(tel)})."
    return None


# --- La prueba: lo que la S13 encontro en la base, aqui NO entra
print("EL PORTERO EN ACCION")
print("=" * 62)
casos_prueba = [
    ("Nombre con numeros",   validar_nombre("Juan123")),
    ("Nombre valido",        validar_nombre("Juan Carlos")),
    ("DNI de 5 digitos",     validar_dni("12345")),
    ("DNI valido",           validar_dni("40506070")),
    ("Correo sin arroba",    validar_correo("juanperezcorreo.com")),
    ("Correo valido",        validar_correo("juan@correo.com")),
    ("Telefono de 3 digitos",validar_telefono("999")),
]
for etiqueta, error in casos_prueba:
    print(f"   {etiqueta:24s} -> {error if error else 'ACEPTADO'}")
print()
print("Los 5 DNI invalidos que encontraste en la S13 entraron porque nadie valido.")
print("Con este portero, no habrian entrado nunca.")
```

**Pregunta 3:** En la Semana 13 encontraste 5 DNI que no tienen 8 dígitos y 11 correos sin arroba. **Ninguno habría entrado con las validaciones de hoy.** Explica la diferencia entre *limpiar* (S13) y *validar* (S14): ¿cuál de las dos es más barata para la empresa y por qué?

Respuesta: Limpiar es un proceso correctivo, en tanto que validar es un proceso preventivo, este último tiene por misión evitar que datos inválidos lleguen a la base, en vez de corregirlos cuando ya estén dentro. Un proceso de limpieza es algo más lento, porque requiere que los datos a corregir hayan entrado primero a la base, lo que puede comprometer la integridad de la misma si los datos se llegan a almacenar o usar antes que sean detectados, por lo que estimo que puede ser más caro para una empresa que una validación; no obstante, creo que ambos procesos deberían de ser parte de la gestión de datos de una empresa.

### Paso 5: la bitácora — que el dato recuerde quién lo tocó

Esta es la pieza que convierte tu sistema en algo **auditable**, y es lo que exige la Ley 29733.

La idea es simple: cada vez que alguien escribe, se guarda una fila en la tabla `auditoria` con **quién** (el rol), **qué** hizo y **cuándo**. Nada más. Tres datos.

Pero fíjate en la consecuencia: cuando el gerente pregunte *"¿quién borró al cliente 47?"*, dejas de responder "no sé" y pasas a responder **"el Supervisor, el 15 de julio a las 14:32"**.

Y ojo con una decisión de diseño que parece pequeña y es enorme: **la bitácora nunca se edita ni se borra**. Solo crece. Una bitácora que se puede modificar no sirve de nada, porque el primero que borraría su rastro es justamente quien hizo algo malo.

```python
# ============================================================
# Paso 5: la bitacora de auditoria (quien, que, cuando)
# ============================================================
def _crear_tabla_auditoria():
    """La bitacora solo CRECE. Nunca se edita ni se borra."""
    c = conn.cursor()
    c.execute("""CREATE TABLE IF NOT EXISTS auditoria (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        fecha   TEXT NOT NULL,
        rol     TEXT NOT NULL,
        accion  TEXT NOT NULL,
        detalle TEXT,
        resultado TEXT)""")
    conn.commit()


def registrar_auditoria(rol, accion, detalle="", resultado="OK"):
    """Deja rastro. Se llama SIEMPRE que alguien escribe, y tambien cuando se le deniega."""
    _crear_tabla_auditoria()
    conn.execute(
        "INSERT INTO auditoria (fecha, rol, accion, detalle, resultado) VALUES (?,?,?,?,?)",
        (datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S"), rol, accion, detalle, resultado))
    conn.commit()


def ver_auditoria(rol, limite=20):
    """Ver la bitacora es un acto de fiscalizacion: solo Auditor y Data Owner."""
    if "Sensible" not in NIVEL_ACCESO.get(rol, set()):
        return f"ACCESO DENEGADO: el rol {rol} no puede ver la bitacora de auditoria."
    _crear_tabla_auditoria()
    return pd.read_sql_query(
        f"SELECT * FROM auditoria ORDER BY id DESC LIMIT {limite}", conn)


_crear_tabla_auditoria()
registrar_auditoria("DataOwner", "Inicio de sesion", "Semana 14: gobierno de datos activado")
print("Bitacora creada. La tabla 'auditoria' ya vive dentro de la base.")
print()
print("Se registran TRES datos por accion: quien (rol), que (accion) y cuando (fecha).")
print("Ojo: tambien se registran los intentos DENEGADOS. Un intento fallido de borrar")
print("datos de salud es justo lo que un auditor quiere ver.")
```

### Paso 6: el CRUD — el sistema por fin trabaja

Ahora juntamos las tres piezas. Cada función de escritura hace lo mismo, **en este orden**:

1. **¿Puede?** → `puede(rol, operacion)` — si no, se deniega **y se registra el intento**
2. **¿Es válido?** → las validaciones — si no, se rechaza sin tocar la base
3. **Escribe** → y recién ahí toca la base
4. **Deja rastro** → `registrar_auditoria(...)`

Ese orden no es casual. Primero el permiso (barato), después la validación (barata), y solo al final la escritura (cara e irreversible).

```python
# ============================================================
# Paso 6: CRUD de clientes -- permiso + validacion + escritura + rastro
# ============================================================
def agregar_cliente(rol, nombres, apellidos, dni, correo, telefono, distrito="Lima"):
    """C de CREATE. Solo Vendedor y DataOwner."""
    # 1. Puede?
    if not puede(rol, "INSERT"):
        registrar_auditoria(rol, "Agregar cliente", f"{nombres} {apellidos}", "DENEGADO")
        return f"ACCESO DENEGADO: el rol {rol} no puede agregar clientes."

    # 2. Es valido? (las reglas de la S13, aplicadas ANTES de grabar)
    for error in (validar_nombre(nombres), validar_nombre(apellidos), validar_dni(dni),
                  validar_correo(correo), validar_telefono(telefono)):
        if error:
            registrar_auditoria(rol, "Agregar cliente", f"{nombres}: {error}", "RECHAZADO")
            return f"DATO INVALIDO: {error}"

    # El DNI no puede repetirse: regla de UNICIDAD (S13)
    existe = pd.read_sql_query(
        "SELECT id_cliente FROM clientes_limpio WHERE num_documento = ?", conn, params=(str(dni),))
    if len(existe):
        registrar_auditoria(rol, "Agregar cliente", f"DNI {dni} duplicado", "RECHAZADO")
        return f"DATO INVALIDO: ya existe un cliente con el DNI {dni}."

    # 3. Escribe
    nid = conn.execute("SELECT COALESCE(MAX(id_cliente),0)+1 FROM clientes_limpio").fetchone()[0]
    cols = pd.read_sql_query("SELECT * FROM clientes_limpio LIMIT 1", conn).columns
    fila = {c: None for c in cols}
    fila.update({"id_cliente": nid, "codigo_cliente": f"C{nid:05d}", "tipo_documento": "Dni",
                 "num_documento": str(dni), "nombres": str(nombres).strip().title(),
                 "apellidos": str(apellidos).strip().title(), "correo": correo,
                 "telefono": str(telefono), "distrito": distrito, "segmento": "Nuevo",
                 "fecha_registro": datetime.date.today().isoformat()})
    if "correo_fue_corregido" in cols:
        fila["correo_fue_corregido"] = 0
    pd.DataFrame([fila]).to_sql("clientes_limpio", conn, if_exists="append", index=False)
    conn.commit()

    # 4. Deja rastro (aquí)
    registrar_auditoria(rol, "Agregar cliente", f"id {nid}: {nombres} {apellidos} (DNI {dni})")
    return f"Cliente {nombres} {apellidos} agregado con id {nid}."


def actualizar_cliente(rol, id_cliente, campo, nuevo_valor):
    """U de UPDATE. Solo Cajero y DataOwner."""
    if not puede(rol, "UPDATE"):
        registrar_auditoria(rol, "Actualizar cliente", f"id {id_cliente}", "DENEGADO")
        return f"ACCESO DENEGADO: el rol {rol} no puede actualizar clientes."

    campos = {"correo": validar_correo, "telefono": validar_telefono,
              "apellidos": validar_nombre, "distrito": validar_nombre}
    campo = str(campo).strip().lower()
    if campo not in campos:
        return f"Campo no valido. Usa: {', '.join(campos)}."
    error = campos[campo](nuevo_valor)
    if error:
        registrar_auditoria(rol, "Actualizar cliente", f"id {id_cliente}: {error}", "RECHAZADO")
        return f"DATO INVALIDO: {error}"

    cur = conn.execute(f"UPDATE clientes_limpio SET {campo}=? WHERE id_cliente=?",
                       (nuevo_valor, int(id_cliente)))
    conn.commit()
    if cur.rowcount == 0:
        return f"No existe el cliente {id_cliente}."
    registrar_auditoria(rol, "Actualizar cliente", f"id {id_cliente}: {campo} = {nuevo_valor}")
    return f"Cliente {id_cliente} actualizado: {campo} = {nuevo_valor}."


def eliminar_cliente(rol, id_cliente):
    """D de DELETE. Solo Supervisor y DataOwner. Es irreversible."""
    if not puede(rol, "DELETE"):
        registrar_auditoria(rol, "Eliminar cliente", f"id {id_cliente}", "DENEGADO")
        return f"ACCESO DENEGADO: el rol {rol} no puede eliminar clientes."

    d = pd.read_sql_query("SELECT nombres, apellidos FROM clientes_limpio WHERE id_cliente=?",
                          conn, params=(int(id_cliente),))
    if not len(d):
        return f"No existe el cliente {id_cliente}."
    quien = f"{d.iloc[0]['nombres']} {d.iloc[0]['apellidos']}"
    conn.execute("DELETE FROM clientes_limpio WHERE id_cliente=?", (int(id_cliente),))
    conn.commit()
    registrar_auditoria(rol, "Eliminar cliente", f"id {id_cliente}: {quien}")
    return f"Cliente {id_cliente} ({quien}) eliminado."


# --- LA PRUEBA
print("=" * 66)
print("1. El VENDEDOR agrega un cliente valido:")
print("  ", agregar_cliente("Vendedor", "Rosa", "Perez", "40506070", "rosa@correo.com", "987654321"))
print()
print("2. El VENDEDOR intenta agregar basura:")
print("  ", agregar_cliente("Vendedor", "Juan123", "Perez", "40506071", "juan@correo.com", "987654322"))
print("  ", agregar_cliente("Vendedor", "Ana", "Torres", "123", "ana@correo.com", "987654323"))
print("  ", agregar_cliente("Vendedor", "Luis", "Diaz", "40506072", "luiscorreo.com", "987654324"))
print()
print("3. El VENDEDOR intenta BORRAR (no puede):")
print("  ", eliminar_cliente("Vendedor", 1))
print()
print("4. El CAJERO corrige un correo (si puede):")
print("  ", actualizar_cliente("Cajero", 2, "correo", "nuevo@correo.com"))
print()
print("5. El AUDITOR intenta escribir (no puede: solo mira):")
print("  ", agregar_cliente("Auditor", "Pedro", "Ruiz", "40506073", "pedro@correo.com", "987654325"))
```

### Paso 7: registrar una venta con su ticket

El vendedor tiene al cliente enfrente. Necesita **vender**, y esa es la operación central del negocio.

Ojo con algo: una venta **no es una fila**. Son tres, en tres tablas: la operación, su detalle y su pago. Si se graba una y falla otra, la base queda inconsistente. Por eso todo va junto.

```python
# ============================================================
# Paso 7: registrar venta + ticket
# ============================================================
def ver_catalogo_productos():
    return pd.read_sql_query("""
        SELECT id_producto, nombre_producto, categoria, precio_base
        FROM productos_servicios WHERE activo = 1
        ORDER BY categoria, nombre_producto""", conn)


def registrar_venta(rol, id_cliente, id_producto, cantidad):
    """Una venta son TRES filas en TRES tablas: operacion, detalle y pago."""
    if not puede(rol, "INSERT"):
        registrar_auditoria(rol, "Registrar venta", f"cliente {id_cliente}", "DENEGADO")
        return f"ACCESO DENEGADO: el rol {rol} no puede registrar ventas."
    try:
        id_cliente, id_producto, cantidad = int(id_cliente), int(id_producto), int(cantidad)
    except ValueError:
        return "DATO INVALIDO: id_cliente, id_producto y cantidad deben ser numeros."
    if cantidad <= 0:
        return "DATO INVALIDO: la cantidad debe ser mayor a 0."

    cli = conn.execute("SELECT nombres, apellidos FROM clientes_limpio WHERE id_cliente=?",
                       (id_cliente,)).fetchone()
    if not cli:
        return f"DATO INVALIDO: no existe el cliente {id_cliente}."
    prod = conn.execute(
        "SELECT nombre_producto, precio_base FROM productos_servicios WHERE id_producto=?",
        (id_producto,)).fetchone()
    if not prod:
        return f"DATO INVALIDO: no existe el producto {id_producto}."

    precio = prod[1]; total = round(precio * cantidad, 2)
    fecha  = datetime.date.today().isoformat()     # SIEMPRE en formato ISO (regla S13)

    nop = conn.execute("SELECT COALESCE(MAX(id_operacion),0)+1 FROM ventas_limpia").fetchone()[0]
    codigo = f"O{nop:05d}"
    cols = pd.read_sql_query("SELECT * FROM ventas_limpia LIMIT 1", conn).columns
    fila = {c: None for c in cols}
    fila.update({"id_operacion": nop, "codigo_operacion": codigo, "id_cliente": id_cliente,
                 "id_sede": 1, "id_empleado": 1, "fecha_operacion": fecha, "canal": "Mostrador",
                 "estado": "Completado", "monto_total": total})
    if "monto_ajustado" in cols:     fila["monto_ajustado"] = total
    if "monto_fue_ajustado" in cols: fila["monto_fue_ajustado"] = 0
    pd.DataFrame([fila]).to_sql("ventas_limpia", conn, if_exists="append", index=False)

    nd = conn.execute("SELECT COALESCE(MAX(id_detalle),0)+1 FROM detalle_operacion").fetchone()[0]
    conn.execute("""INSERT INTO detalle_operacion
        (id_detalle, id_operacion, id_producto, cantidad, precio_unitario, subtotal)
        VALUES (?,?,?,?,?,?)""", (nd, nop, id_producto, cantidad, precio, total))

    npg = conn.execute("SELECT COALESCE(MAX(id_pago),0)+1 FROM pagos").fetchone()[0]
    conn.execute("""INSERT INTO pagos
        (id_pago, codigo_pago, id_operacion, fecha_pago, medio_pago, monto_pagado, estado_pago)
        VALUES (?,?,?,?,?,?,?)""",
        (npg, f"PG{npg:05d}", nop, fecha, "Efectivo", total, "Pagado"))
    conn.commit()

    registrar_auditoria(rol, "Registrar venta", f"{codigo}: cliente {id_cliente}, S/. {total}")

    return "\n".join(["=" * 46, "        FARMACIA MEDISUR - TICKET DE VENTA", "=" * 46,
        f"Codigo venta : {codigo}", f"Fecha        : {fecha}",
        f"Cliente      : {cli[0]} {cli[1]}", f"Atendido por : {rol}", "-" * 46,
        f"Producto     : {prod[0]}", f"Cantidad     : {cantidad}",
        f"Precio unit. : S/. {precio:.2f}", f"TOTAL        : S/. {total:.2f}",
        "=" * 46, "         Gracias por su compra!"])


def ver_historial_ventas(rol, limite=8):
    if not puede(rol, "SELECT"):
        return f"ACCESO DENEGADO: el rol {rol} no puede consultar."
    return pd.read_sql_query(f"""
        SELECT v.codigo_operacion, v.fecha_operacion,
               c.nombres || ' ' || c.apellidos AS cliente, v.monto_total, v.estado
        FROM ventas_limpia v JOIN clientes_limpio c ON v.id_cliente = c.id_cliente
        ORDER BY v.id_operacion DESC LIMIT {limite}""", conn)


print("--- CATALOGO (que puede comprar el cliente) ---")
display(ver_catalogo_productos().head(4))
print("--- EL VENDEDOR REGISTRA UNA VENTA ---")
print(registrar_venta("Vendedor", 2, 3, 2))
print()
print("--- EL AUDITOR INTENTA VENDER (no puede) ---")
print(registrar_venta("Auditor", 2, 3, 1))
```

Eso es a propósito: ocultar la opción esconde el problema; denegarla y registrarla lo documenta.

```python
# ============================================================
# Paso 9: EL MENU de la Semana 14
# ============================================================
ROLES_MENU = {"1":"Vendedor","2":"Cajero","3":"Supervisor","4":"Auditor","5":"DataOwner"}

def _mostrar(r):
    display(r) if isinstance(r, pd.DataFrame) else print(r)


def _descargar(ruta):
    try:
        from google.colab import files; files.download(ruta)
        import time; time.sleep(1.5)   # pausa: sin ella el navegador bloquea descargas seguidas
        print("  Descargado:", ruta)
    except Exception:
        print("  Guardado en el entorno:", ruta)


def _exportar_tabla(d, nombre, con_indice=False):
    """Guarda la tabla en CSV y Excel, y descarga los dos archivos."""
    if input("  Descargar esta tabla en CSV y Excel? (s/n): ").strip().lower() != "s":
        return
    ruta_csv, ruta_xlsx = nombre + ".csv", nombre + ".xlsx"
    d.to_csv(ruta_csv, index=con_indice, encoding="utf-8-sig")
    try:
        d.to_excel(ruta_xlsx, index=con_indice)
    except Exception as e:
        print("  No se pudo crear el Excel:", e); ruta_xlsx = None
    _descargar(ruta_csv)
    if ruta_xlsx: _descargar(ruta_xlsx)


def submenu_clientes(rol):
    while True:
        print("\n--- CLIENTES ---")
        print("  1) Ver lista     2) Agregar     3) Actualizar     4) Eliminar")
        print("  0) Volver")
        op = input("  Opcion: ").strip()
        if op == "0": break
        elif op == "1":
            _mostrar(ver_clientes(rol))
            print(f"  (Como {rol} ves {len(columnas_visibles('clientes_limpio', rol))} columnas.)")
        elif op == "2":
            print(_mostrar(agregar_cliente(rol, input("  Nombres: "), input("  Apellidos: "),
                                           input("  DNI: "), input("  Correo: "),
                                           input("  Telefono: "))))
        elif op == "3":
            print(_mostrar(actualizar_cliente(rol, input("  id_cliente: "),
                                              input("  Campo (correo/telefono/apellidos/distrito): "),
                                              input("  Nuevo valor: "))))
        elif op == "4":
            print(_mostrar(eliminar_cliente(rol, input("  id_cliente: "))))
        else: print("  Opcion invalida.")


def submenu_ventas(rol):
    while True:
        print("\n--- VENTAS ---")
        print("  1) Ver catalogo   2) Registrar venta + ticket   3) Historial")
        print("  0) Volver")
        op = input("  Opcion: ").strip()
        if op == "0": break
        elif op == "1": _mostrar(ver_catalogo_productos())
        elif op == "2":
            _mostrar(ver_catalogo_productos().head(10))
            print(registrar_venta(rol, input("  id_cliente: "), input("  id_producto: "),
                                  input("  cantidad: ")))
        elif op == "3": _mostrar(ver_historial_ventas(rol))
        else: print("  Opcion invalida.")


def submenu_gobierno(rol):
    while True:
        print("\n--- GOBIERNO Y AUDITORIA ---")
        print("  1) Matriz de gobierno (Data Owner / Steward)")
        print("  2) Matriz de permisos por rol")
        print("  3) Bitacora de auditoria        (Auditor / Data Owner)")
        print("  4) Intentos bloqueados          (Auditor / Data Owner)")
        print("  0) Volver")
        op = input("  Opcion: ").strip()
        if op == "0": break
        elif op == "1":
            g = pd.read_sql_query(
                "SELECT tabla, columna, sensibilidad, data_owner, data_steward FROM matriz_gobierno", conn)
            _mostrar(g)
            _exportar_tabla(g, "matriz_gobierno")
        elif op == "2":
            p = pd.read_sql_query("SELECT * FROM matriz_permisos", conn)
            _mostrar(p)
            _exportar_tabla(p, "matriz_permisos")
        elif op == "3":
            b = ver_auditoria(rol)
            _mostrar(b)
            if isinstance(b, pd.DataFrame):
                _exportar_tabla(b, "bitacora_auditoria")
        elif op == "4":
            if "Sensible" not in NIVEL_ACCESO.get(rol, set()):
                print(f"  ACCESO DENEGADO: el rol {rol} no puede ver la auditoria.")
            else:
                _mostrar(pd.read_sql_query(
                    "SELECT fecha, rol, accion, detalle, resultado FROM auditoria "
                    "WHERE resultado IN ('DENEGADO','RECHAZADO') ORDER BY id DESC LIMIT 15", conn))
        else: print("  Opcion invalida.")


def menu():
    print("=" * 56)
    print("        SISTEMA DE DATOS - FARMACIA MEDISUR")
    print("           Semana 14: gobierno de datos")
    print("=" * 56)
    print("Elige tu rol:  1) Vendedor  2) Cajero  3) Supervisor")
    print("               4) Auditor   5) Data Owner")
    rol = ROLES_MENU.get(input("Numero de tu rol: ").strip())
    if rol is None:
        print("Rol no valido."); return
    print(f"\nBienvenido, {rol}.")
    print(f"  Puedes ver   : {', '.join(sorted(NIVEL_ACCESO[rol]))}")
    print(f"  Puedes hacer : {', '.join(sorted(PERMISOS[rol]))}")
    registrar_auditoria(rol, "Inicio de sesion", "menu()")

    while True:
        print("\n" + "=" * 56)
        print(f"MENU PRINCIPAL     (rol: {rol})")
        print("=" * 56)
        print("  1) Clientes  (ver, agregar, actualizar, eliminar)")
        print("  2) Ventas    (catalogo, registrar + ticket, historial)   <-- NUEVO")
        print("  3) Catalogo de datos (metadatos)")
        print("  4) Calidad de datos")
        print("  5) Gobierno y auditoria                                  <-- NUEVO")
        print("  6) Descargar la base limpia (.db)")
        print("  0) Salir")
        op = input("Opcion: ").strip()
        if op == "0":
            registrar_auditoria(rol, "Cierre de sesion")
            print("Sesion cerrada. Hasta luego,", rol); break
        elif op == "1": submenu_clientes(rol)
        elif op == "2": submenu_ventas(rol)
        elif op == "3":
            _mostrar(pd.read_sql_query(
                "SELECT tabla, columna, sensibilidad FROM catalogo_metadatos LIMIT 15", conn))
        elif op == "4":
            print("  (El scorecard de la Semana 13 sigue disponible en su notebook.)")
        elif op == "5": submenu_gobierno(rol)
        elif op == "6": _descargar(DB_DESTINO)
        else: print("Opcion invalida.")


print("Menu de la Semana 14 listo. Ejecuta  menu()  en la siguiente celda.")
```
